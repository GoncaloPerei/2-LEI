# Polimorfismo

**Polimorfismo:**

- É considerado "Qualidade ou estado de ser capaz de assumir diferentes formas"
- É caracterizado quando duas ou mais classes distintas têm métodos de mesmo nome

<br>
Desta forma, uma função poderá usar um objeto de qualquer uma das classes polimórfica, **sem necessidade de tratar de forma diferenciada** conforme a classe do objeto

- O polimorfismo permite que referências de tipos de classes mais abstratas representem o comportamento das classes concretas
- Os objetos possuem a capacidade de assumir diferentes formas

<br>
<br>
Exemplo:
- Listas ou Strings podem ser indexadas por um número e usam a função len()

- Um algoritmo pode ser aplicado a um objeto X e também aplicado a um objeto Y.
```python
for i in range(len(X)):
    print(i, X[i])
```

### Formas de polimorfismo

Formas de implementar polimorfismo:
- Duck Typing
- Sobrecarga de método
- Sobrecarga de operador
- Substituição de método

### Duck Typing

Permite definir o tipo de variável/objecto em tempo de execução

- Não é necessário mencionar explicitamente o tipo do objeto antes de realizar qualquer tipo de operação no esse objeto
- Isto é possível devido ao conceito de tipagem dinâmica

Exemplo:

````python
class Duck:
    def swim(self):
        print("The duck is swimming")

    def fly(self):
        print("The duck is flying")

class Swan:
    def swim(self):
        print("The swan is swimming")

    def fly(self):
        print("The swan is flying")

class Penguin:
    def swim(self):
        print("The penguin is swimming")

birds = [Duck(), Swan(), Penguin()]

for bird in birds:
  bird.swim()
````

### Substituição de método

Permite reutilizar código para outra classe.

- Corresponde a um mecanismo que utiliza a herança
- Basta definir a classe filha em um método com o mesmo nome de um método na classe base


````python
class Programacao:
    Datastructure = True
    Algorithms = True

    def praticar(self):
        print('Praticar torna aluno perfeito')

    def consistencia(self):
        print('Trabalho duro com consistencia aumenta resultados')

class POO(Programacao):
    def consistencia(self):
        print('Trabalho duro com consistencia aumenta resultados_2')


poo = POO()
poo.consistencia()
````

### Method overloading (Sobrecarga de método)

Permite chamar o mesmo método com parâmetros diferentes.
<br>
Python não suporta sobrecarga de método, considera apenas o último método definido, neste sentido precisamos escrever a lógica do método para que diferentes códigos sejam executados dentro da função dependendo do parâmetro passado.

<br>
exemplo:

````python
## Polimorfismo por method overloading

class Shape:

    def area(self, a, b=0):
        if b > 0:
            print('area do rectangle é: ', a*b)
        else:
            print('area do square: ', a**2)


square = Shape()
square.area(5)

rectangle = Shape()
rectangle.area(5,3)
````

### Operator overloading (Sobrecarga de operador)

Permite alterar o comportamento padrão de um operador dependendo dos operandos (valores) que usamos. Ou seja, podemos usar o mesmo operador para vários propósitos.

- Por exemplo: operador + executará uma operação de adição aritmética quando usado com números. Tal como fará a concatenação quando usado com strings.


```python
# add numbers
print(100 +200)

# concatenate 2 strings
print('coca' + 'cola')

# join 2 lists
print([10, 20] + ['coca', 'cola'])
```


````python
class Book:
    def __init__(self,pages):
        self.pages = pages

    # Overloading '+' operator with magic methods
    def __add__(self,other):
        return self.pages + other.pages

book1 = Book(200)
book2 = Book(300)

print(book1 + book2)
````

<br>
O operador *>* nos diz se o primeiro operando é maior do que o segundo

O método *__gt__*, é a abreviatura de maior do que; este método deve devolver o resultado correcto da comparação.
- deve retornar True se e só se o objecto actual for maior do que o objecto passado como argumento.  - Os critérios utilizados podem ser determinados pelo programador.
- Por objecto actual, entendemos o objecto sobre o qual o método é chamado com a notação dot "."

````python
class Product:
    def __init__(self, name: str, price: float):
        self.__name = name
        self.__price = price

    def __str__(self):
        return f"{self.__name} (price {self.__price})"

    @property
    def price(self):
        return self.__price

    def __gt__(self, another_product):
        return self.price > another_product.price


orange = Product("Orange", 2.90)
apple = Product("Apple", 3.95)

if orange > apple:
    print("Orange is greater")
else:
    print("Apple is greater")
````

````python
class Product:
    def __init__(self, name: str, price: float):
        self.__name = name
        self.__price = price

    def __str__(self):
        return f"{self.__name} (price {self.__price})"

    @property
    def price(self):
        return self.__price

    @property
    def name(self):
        return self.__name

    def __gt__(self, another_product):
        return self.name > another_product.name
````

````python
Orange = Product("Orange", 4.90)
Apple = Product("Apple", 3.95)

if Orange > Apple:
    print("Orange is greater")
else:
    print("Apple is greater")
````

### Outros operadores


Exemplo:

````python
from datetime import datetime

class Note:
    def __init__(self, entry_date: datetime, entry: str):
        self.entry_date = entry_date
        self.entry = entry

    def __str__(self):
        return f"{self.entry_date}: {self.entry}"

    def __add__(self, another):
        # The date of the new note is the current time
        new_note = Note(datetime.now(), "")
        new_note.entry = self.entry + " and " + another.entry
        return new_note


entry1 = Note(datetime(2016, 12, 17), "Remember to buy presents")
entry2 = Note(datetime(2016, 12, 23), "Remember to get a tree")

# These notes can be added together with the + operator
# This calls the  __add__ method in the Note class
both = entry1 + entry2
print(both)
````
