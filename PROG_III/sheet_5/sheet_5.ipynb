```python
class RacingSeries:
    """Base class representing a motorsport racing championship series.

    Attributes:
        name (str): Name of the racing series.
        drivers (list[str]): List of participating driver names.
        points (dict[str, int]): Dictionary mapping driver names to their total points.
    """

    def __init__(self, name: str, drivers: list[str], points: dict[str, int] | None = None):
        """Initializes a new RacingSeries instance.

        Args:
            name (str): Name of the racing series.
            drivers (list[str]): List of participating driver names.
            points (dict[str, int] | None): Optional dictionary mapping drivers to
                initial points. Defaults to None, initializing all drivers with 0 points.
        """
        self.name = name
        self.drivers = drivers
        self.points = points if points is not None else {driver: 0 for driver in drivers}

    # -------------------------------------------------------------------------
    # Public Methods
    # -------------------------------------------------------------------------
    def add_race_result(self, finishing_order: list[str]):
        """Updates championship points based on a race's finishing order.

        Args:
            finishing_order (list[str]): Ordered list of driver names from 1st place
                to last place.
        """
        for position, driver in enumerate(finishing_order, start=1):
            if driver in self.points:
                self.points[driver] += self.get_points(position)

    def get_points(self, position: int) -> int:
        """Calculates points awarded for a given finishing position.

        Intended to be overridden by subclasses implementing specific scoring rules.

        Args:
            position (int): Finishing position in the race (1-indexed).

        Returns:
            int: Points awarded for the position (defaults to 0 in base class).
        """
        return 0

    def get_standings(self) -> list[tuple[str, int]]:
        """Retrieves the current championship standings sorted by points.

        Returns:
            list[tuple[str, int]]: List of (driver_name, total_points) tuples sorted
                in descending order by points.
        """
        return sorted(self.points.items(), key=lambda x: x[1], reverse=True)

    def display_standings(self):
        """Prints the current championship standings in a human-readable format."""
        print(f"\n--- Classificação: {self.name} ---")
        for pos, (driver, pts) in enumerate(self.get_standings(), start=1):
            print(f"{pos}º lugar: {driver} - {pts} pts")


class FormulaSeries(RacingSeries):
    """Represents a Formula 1 championship series using the official F1 points system."""

    # -------------------------------------------------------------------------
    # Overridden Methods
    # -------------------------------------------------------------------------
    def get_points(self, position: int) -> int:
        """Calculates points using the Formula 1 scoring system (top 10 score).

        Args:
            position (int): Finishing position in the race (1-indexed).

        Returns:
            int: Points awarded (25 down to 1 for positions 1 to 10, 0 otherwise).
        """
        tabela_f1 = {
            1: 25, 2: 18, 3: 15, 4: 12, 5: 10,
            6: 8,  7: 6,  8: 4,  9: 2,  10: 1
        }
        return tabela_f1.get(position, 0)


class MotoGPSeries(RacingSeries):
    """Represents a MotoGP championship series using the official MotoGP points system."""

    # -------------------------------------------------------------------------
    # Overridden Methods
    # -------------------------------------------------------------------------
    def get_points(self, position: int) -> int:
        """Calculates points using the MotoGP scoring system (top 15 score).

        Args:
            position (int): Finishing position in the race (1-indexed).

        Returns:
            int: Points awarded (25 down to 1 for positions 1 to 15, 0 otherwise).
        """
        tabela_motogp = {
            1: 25, 2: 20, 3: 16, 4: 13, 5: 11,
            6: 10, 7: 9,  8: 8,  9: 7,  10: 6,
            11: 5, 12: 4, 13: 3, 14: 2, 15: 1
        }
        return tabela_motogp.get(position, 0)


class VintageSeries(RacingSeries):
    """Represents a vintage racing series using a linear decreasing points system."""

    # -------------------------------------------------------------------------
    # Overridden Methods
    # -------------------------------------------------------------------------
    def get_points(self, position: int) -> int:
        """Calculates points using a linear formula (11 - position, minimum 0).

        Args:
            position (int): Finishing position in the race (1-indexed).

        Returns:
            int: Points awarded (10 for 1st down to 1 for 10th, 0 otherwise).
        """
        return max(0, 11 - position)


# -----------------------------------------------------------------------------
# Test Execution & Championship Simulation
# -----------------------------------------------------------------------------

# Initialize driver rosters (minimum 5 drivers per series)
pilotos_f1 = ["Max Verstappen", "Lando Norris", "Charles Leclerc", "Lewis Hamilton", "Oscar Piastri"]
pilotos_motogp = ["Jorge Martín", "Francesco Bagnaia", "Marc Márquez", "Enea Bastianini", "Pedro Acosta"]
pilotos_vintage = ["Ayrton Senna", "Alain Prost", "Niki Lauda", "Nelson Piquet", "Nigel Mansell"]

# Instantiate each racing series
f1 = FormulaSeries("Campeonato Mundial de F1", pilotos_f1)
motogp = MotoGPSeries("Campeonato Mundial de MotoGP", pilotos_motogp)
vintage = VintageSeries("Taça Clássica Vintage", pilotos_vintage)

# Record Formula 1 race results (3 races)
f1.add_race_result(["Max Verstappen", "Lando Norris", "Charles Leclerc", "Oscar Piastri", "Lewis Hamilton"])
f1.add_race_result(["Lando Norris", "Max Verstappen", "Lewis Hamilton", "Charles Leclerc", "Oscar Piastri"])
f1.add_race_result(["Charles Leclerc", "Oscar Piastri", "Lando Norris", "Max Verstappen", "Lewis Hamilton"])

# Record MotoGP race results (3 races)
motogp.add_race_result(["Francesco Bagnaia", "Jorge Martín", "Marc Márquez", "Enea Bastianini", "Pedro Acosta"])
motogp.add_race_result(["Jorge Martín", "Marc Márquez", "Francesco Bagnaia", "Pedro Acosta", "Enea Bastianini"])
motogp.add_race_result(["Marc Márquez", "Jorge Martín", "Enea Bastianini", "Francesco Bagnaia", "Pedro Acosta"])

# Record Vintage Series race results (3 races)
vintage.add_race_result(["Ayrton Senna", "Alain Prost", "Nelson Piquet", "Nigel Mansell", "Niki Lauda"])
vintage.add_race_result(["Alain Prost", "Ayrton Senna", "Niki Lauda", "Nigel Mansell", "Nelson Piquet"])
vintage.add_race_result(["Ayrton Senna", "Nigel Mansell", "Alain Prost", "Nelson Piquet", "Niki Lauda"])

# Display final standings for all series
f1.display_standings()
motogp.display_standings()
vintage.display_standings()
```
