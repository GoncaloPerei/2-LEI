```python
class Book:
    """Base class representing a standard book in a library system.

    Attributes:
        isbn (str): Unique International Standard Book Number (read-only).
        title (str): Title of the book.
        author (str): Author of the book.
        available_copies (int): Number of copies currently available for borrowing.
        total_copies (int): Total number of copies owned by the library.
        borrowed_by (list[str]): List of member names currently borrowing the book.
    """

    def __init__(self, isbn: str, title: str, author: str, total_copies: int):
        """Initializes a new Book instance.

        Args:
            isbn (str): Unique ISBN identifier for the book.
            title (str): Title of the book.
            author (str): Author of the book.
            total_copies (int): Total number of copies available initially.

        Raises:
            ValueError: If isbn is empty or total_copies is negative.
        """
        if not isinstance(isbn, str) or not isbn.strip():
            raise ValueError("ISBN cannot be empty.")
        if not isinstance(total_copies, int) or total_copies < 0:
            raise ValueError("Total copies must be a non-negative integer.")
        
        self.__isbn = isbn.strip()
        self.__title = title
        self.__author = author
        self.__available_copies = total_copies
        self.__total_copies = total_copies
        self.__borrowed_by: list[str] = []

    # -------------------------------------------------------------------------
    # Properties & Setters
    # -------------------------------------------------------------------------
    @property
    def isbn(self) -> str:
        return self.__isbn

    @property
    def title(self) -> str:
        return self.__title

    @title.setter
    def title(self, value: str) -> None:
        if not isinstance(value, str):
            raise TypeError("Title must be a string")
        if not value.strip():
            raise TypeError("Title cannot be empty")
        self.__title = value.strip()

    @property
    def author(self) -> str:
        return self.__author

    @author.setter
    def author(self, value: str) -> None:
        if not isinstance(value, str):
            raise TypeError("Author must be a string")
        if not value.strip():
            raise TypeError("Author cannot be empty")
        self.__author = value.strip()

    @property
    def available_copies(self) -> int:
        return self.__available_copies

    @property
    def total_copies(self) -> int:
        return self.__total_copies

    @property
    def borrowed_by(self) -> list[str]:
        return self.__borrowed_by.copy()

    # -------------------------------------------------------------------------
    # Public Methods
    # -------------------------------------------------------------------------
    def borrow(self, member_name: str) -> None:
        """Registers a book loan to a library member.

        Args:
            member_name (str): Name of the member borrowing the book.

        Raises:
            ValueError: If member_name is empty or no copies are available.
        """
        if not isinstance(member_name, str) or not member_name.strip():
            raise ValueError("Member name cannot be empty.")
        if self.__available_copies <= 0:
            raise ValueError("This book doesn't have available copies.")

        self.__borrowed_by.append(member_name.strip())
        self.__available_copies -= 1

    def return_book(self, member_name: str) -> None:
        """Registers the return of a borrowed book from a library member.

        Args:
            member_name (str): Name of the member returning the book.

        Raises:
            ValueError: If the specified member does not currently have the book.
        """
        try:
            self.__borrowed_by.remove(member_name)
        except ValueError:
            raise ValueError(f"This book wasn't borrowed by {member_name}.") from None
        
        self.__available_copies += 1

    def is_available(self) -> bool:
        """Checks whether at least one copy of the book is available.

        Returns:
            bool: True if available_copies is greater than 0, False otherwise.
        """
        return self.__available_copies > 0

    def display_book_info(self) -> None:
        status = "Available" if self.is_available() else "Unavailable"
        borrowers = ", ".join(self.__borrowed_by) if self.__borrowed_by else "None"

        print(f"Title: {self.__title}")
        print(f"Author: {self.__author}")
        print(f"ISBN: {self.__isbn}")
        print(f"Copies: {self.__available_copies}/{self.__total_copies} ({status})")
        print(f"Borrowed by: {borrowers}")


class ReferenceBook(Book):
    """Represents a reference book restricted to in-library consultation only.

    Attributes:
        location (str): Physical shelf or room location inside the library.
        consultations (list[str]): History of members who consulted the book.
    """

    def __init__(self, isbn: str, title: str, author: str, total_copies: int, location: str):
        """Initializes a new ReferenceBook instance.

        Args:
            isbn (str): Unique ISBN identifier for the book.
            title (str): Title of the book.
            author (str): Author of the book.
            total_copies (int): Total number of copies owned by the library.
            location (str): Physical location of the book inside the library.
        """
        super().__init__(isbn, title, author, total_copies)
        self.__location_in_library = location
        self.__consultations: list[str] = []

    # -------------------------------------------------------------------------
    # Properties & Setters
    # -------------------------------------------------------------------------
    @property
    def location(self) -> str:
        return self.__location_in_library

    @location.setter
    def location(self, value: str) -> None:
        if not isinstance(value, str):
            raise TypeError("Location must be a string")
        if not value.strip():
            raise TypeError("Location cannot be empty")
        self.__location_in_library = value.strip()

    @property
    def consultations(self) -> list[str]:
        return self.__consultations.copy()

    # -------------------------------------------------------------------------
    # Public & Overridden Methods
    # -------------------------------------------------------------------------
    def borrow(self, member_name: str) -> None:
        """Overrides base borrow method to prevent external loans.

        Args:
            member_name (str): Name of the member attempting to borrow the book.

        Raises:
            ValueError: Always raised because reference books cannot be borrowed.
        """
        raise ValueError(f"Reference books ('{self.title}') cannot be borrowed; they are for library consultation only.")

    def consult_in_library(self, member_name: str) -> None:
        """Logs an in-library consultation by a member.

        Args:
            member_name (str): Name of the member consulting the book.

        Raises:
            ValueError: If member_name is empty.
        """
        if not isinstance(member_name, str) or not member_name.strip():
            raise ValueError("Member name cannot be empty.")
        self.__consultations.append(member_name.strip())

    def display_book_info(self) -> None:
        super().display_book_info()
        consulted_by = ", ".join(self.__consultations) if self.__consultations else "None"
        
        print(f"Type: Reference Book (In-library consultation only)")
        print(f"Location: {self.__location_in_library}")
        print(f"Consultations: {len(self.__consultations)} (Consulted by: {consulted_by})")


# -----------------------------------------------------------------------------
# Test Execution & Library Simulation
# -----------------------------------------------------------------------------

# Instantiate standard and reference books
print("=== 1. CRIAR LIVROS ===")
book1 = Book("978-972-0-04957-5", "Os Maias", "Eça de Queirós", 3)
book2 = Book("978-989-623-189-7", "Ensaio sobre a Cegueira", "José Saramago", 1)
ref_book1 = ReferenceBook("978-972-20-3456-2", "Dicionário da Língua Portuguesa", "Porto Editora", 2, "Piso 1 - Estante A3")
ref_book2 = ReferenceBook("978-0-13-235088-4", "Clean Code (Edição de Arquivo)", "Robert C. Martin", 1, "Sala de Leitura - Secção TI")

# Display initial book details
book1.display_book_info()
print("-" * 40)
ref_book1.display_book_info()

# Test standard borrowing and returning workflow
print("\n=== 2. TESTAR EMPRÉSTIMOS E DEVOLUÇÕES (Book) ===")
book1.borrow("Ana Silva")
book1.borrow("João Santos")
print(f"Cópias disponíveis de '{book1.title}': {book1.available_copies}")

book1.return_book("Ana Silva")
print(f"Após devolução da Ana: {book1.available_copies} cópias | Emprestado a: {book1.borrowed_by}")

# Test in-library consultation logging for reference books
print("\n=== 3. TESTAR CONSULTAS (ReferenceBook) ===")
ref_book1.consult_in_library("Mariana Costa")
ref_book1.consult_in_library("Pedro Alves")
ref_book1.display_book_info()

# Test business rules, validations, and expected exceptions
print("\n=== 4. TESTAR VALIDAÇÕES E ERROS ESPERADOS ===")

# Attempt to borrow a reference book (should fail)
try:
    ref_book1.borrow("Carlos Ferreira")
except ValueError as e:
    print(f"[OK - Bloqueado] {e}")

# Attempt to borrow when no copies remain (should fail)
try:
    book2.borrow("Inês Martins")
    book2.borrow("Rui Oliveira")
except ValueError as e:
    print(f"[OK - Sem stock] {e}")

# Attempt to return a book not borrowed by the given member (should fail)
try:
    book1.return_book("Pessoa Desconhecida")
except ValueError as e:
    print(f"[OK - Devolução inválida] {e}")

# Attempt to assign an empty title via property setter (should fail)
try:
    book1.title = "   "
except (ValueError, TypeError) as e:
    print(f"[OK - Setter bloqueado] {e}")
```
