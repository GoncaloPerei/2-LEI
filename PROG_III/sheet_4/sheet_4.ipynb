### **Exercise 1.**
Create a class **"Lamp"** and add methods to turn the lamp on/off.
Attributes:
 - Division: Name of the division where the lamp is located.
 - On: Indicates if lamp is turned on. Initial state should be *False*.


```python
class Lamp:
    """Represents an lighting fixture with controlled state and division placement.

    Attributes:
        division (str): Name of the room/division where the lamp is located.
        on (bool): Operational status of the lamp (True for on, False for off).
    """
    def __init__(self, division: str, on: bool = False):
        """Initializes a new Lamp instance.

        Delegates attribute assignment directly to properties to enforce
        type and value validation during initialization.
        """
        self.__division = division
        self.__on = on

    # -------------------------------------------------------------------------
    # Properties: division
    # -------------------------------------------------------------------------
    @property
    def division(self) -> str:
        return self.__division

    @division.setter
    def division(self, value: str):
        """Sets and validates the room/division name.

        Raises:
            TypeError: If value is not a string.
            ValueError: If value is empty or consists solely of whitespace.
        """
        if not isinstance(value, str):
            raise TypeError("Division must be a string.")
        if not value.strip():
            raise ValueError("Division cannot be empty.")
        self.__division = value.strip()

    # -------------------------------------------------------------------------
    # Properties: on (state)
    # -------------------------------------------------------------------------
    @property
    def on(self) -> bool:
        return self.__on
    
    @on.setter
    def on(self, value: bool):
        """Sets and validates the lamp's power status.

        Raises:
            TypeError: If the value provided is not a boolean.
        """
        if not isinstance(value, (bool)):
            raise ValueError("Value must be a boolean (True or False).")
        
        self.__on = bool(value)
    
    def turn_on(self):
        self.on = True
    
    def turn_off(self):
        self.on = False

    def description(self):
        """Generates a string for status description of the lamp.

        Returns:
            str: Description formatted as '<Division>: Turned On' or '<Division>: Turned Off'.
        """
        status = "Turned On" if self.on else "Turned Off"
        return f"{self.division}: {status}"

living_room= Lamp("Living Room")
bedroom = Lamp("Bedroom")

living_room.turn_on()

assert living_room.on is True
assert bedroom.on is False
assert living_room.description() == "Living Room: Turned On"

living_room.turn_off()
assert living_room.description() == "Living Room: Turned Off"
```

### **Exercise 2.**

Create a class **DigitalWallet** and add methods to deposit, pay and check budget on your wallet. Attributes:
 - Holder: Name of the *Wallet* holder.
 - Balance: Value *available* on the wallet, initially zero. 

```python
from decimal import Decimal

class Wallet:
    """Represents a financial wallet with strict monetary precision and balance tracking.

    Uses `decimal.Decimal` to avoid floating-point rounding errors in monetary operations.

    Attributes:
        holder (str): Name of the wallet owner.
        balance (Decimal): Current available balance (non-negative).
    """
    def __init__(self, holder: str, balance: Decimal = Decimal("0.00")):
        """Initializes a new Wallet instance.

        Delegates assignments to property setters to enforce validation rules
        (type checks, non-empty holder, and non-negative initial balance).
        """
        self.__holder = holder
        self.__balance = balance

    # -------------------------------------------------------------------------
    # Properties: holder
    # -------------------------------------------------------------------------
    @property
    def holder(self) -> str:
        return self.__holder

    @holder.setter
    def holder(self, value: str):
        """Sets and validates the wallet holder's name.

        Raises:
            TypeError: If the holder is not a string.
            ValueError: If the holder string is empty or contains only whitespace.
        """
        if not isinstance(value, str):
            raise TypeError("Holder must be a string.")
        if not value.strip():
            raise ValueError("Holder cannot be empty.")
        self.__holder = value.strip()

    # -------------------------------------------------------------------------
    # Properties: balance
    # -------------------------------------------------------------------------
    @property
    def balance(self) -> Decimal:
        return self.__balance

    @balance.setter
    def balance(self, value: Decimal):
        """Sets and validates the wallet balance.

        Raises:
            TypeError: If the value provided is not an instance of Decimal.
            ValueError: If the value is strictly negative.
        """
        if type(value) != Decimal:
            raise TypeError("Balance must be a decimal.")
        if value < 0:
            raise ValueError("Cannot set negative balance.")
        self.__balance = value

    # -------------------------------------------------------------------------
    # Public Methods
    # -------------------------------------------------------------------------
    def deposit(self, value: Decimal):
        """Deposits a positive monetary amount into the wallet.

        Args:
            value (Decimal): Amount to add to the current balance.

        Raises:
            TypeError: If value is not a Decimal.
            ValueError: If value is not strictly greater than zero.
        """
        if type(value) != Decimal:
            raise TypeError("Deposit must be a decimal.")
        if value <= 0:
            raise ValueError("The deposit must be positive.")
        self.balance += value

    def pay(self, value: Decimal):
        """Deducts an expense amount from the available balance.

        Args:
            value (Decimal): Amount to pay.

        Raises:
            TypeError: If value is not a Decimal.
            ValueError: If value is non-positive or exceeds current funds.
        """
        if type(value) != Decimal:
            raise TypeError("Payment must be a decimal.")
        if value <= 0:
            raise ValueError("The payment must be positive.")
        if value > self.balance:
            raise ValueError("Insufficient balance.")
        self.balance -= value

wallet = Wallet("Goncalo", Decimal("100.00"))

wallet.deposit(Decimal("50"))
wallet.pay(Decimal("30"))

assert wallet.balance == Decimal("120")

try:
    wallet.pay(Decimal("200"))
except ValueError:
    print("Payment Refused.")
else:
    raise AssertionError("Payment should be refused.")

assert wallet.balance == Decimal("120")
        
```

### **Exercise 3.**

Create a class **ClassSchool** and add methods to add students and to check the number of students. Attributes:
 - Name: Name of the class.
 - Students: List of the students of the class.

```python
class ClassSchool:
    """Represents a school classroom/grade unit within an educational institution.

    Attributes:
        school (str): Class-level attribute shared across all class instances.
        name (str): Unique class identifier or grade label (e.g., 'A', '10-B').
        students (list[str]): List of enrolled student names.
    """
    school: str = "ABC School"

    def __init__(self, name: str, students: list | None = None):
        """Initializes a new ClassSchool instance.

        Args:
            name (str): Identifier for the school class.
            students (list | None): Optional initial list of student names.
                Defaults to None, creating an independent empty list.

        Raises:
            TypeError: If the class name is not a string.
            ValueError: If the class name is empty or only whitespace.
        """
        self.__name = name
        self.__students = students if students is not None else []

    # -------------------------------------------------------------------------
    # Properties: name
    # -------------------------------------------------------------------------
    @property
    def name(self) -> str:
        return self.__name

    @name.setter
    def name(self, value: str):
        """Sets and validates the class name.

        Raises:
            TypeError: If the value is not a string.
            ValueError: If the string is empty or contains only whitespace.
        """
        if not isinstance(value, str):
            raise TypeError("Holder must be a string.")
        if not value.strip():
            raise ValueError("Holder cannot be empty.")
        self.__name = value.strip()

    # -------------------------------------------------------------------------
    # Properties: students
    # -------------------------------------------------------------------------
    @property
    def students(self) -> list:
        return list(self.__students)

    # -------------------------------------------------------------------------
    # Public Methods
    # -------------------------------------------------------------------------
    def addStudent(self, name: str):
        """Enrolls a new student into the class.

        Args:
            name (str): Name of the student.

        Raises:
            TypeError: If student name is not a string.
            ValueError: If student name is empty or only whitespace.
        """
        if type(name) != str:
            raise TypeError("Student name must be a string.")
        if not name.strip():
            raise ValueError("Student name cannot be empty.")
        self.__students.append(name.strip())

    def numberStudents(self):
        """Calculates the total number of enrolled students.

        Returns:
            int: Student count.
        """
        return len(self.students)

class_a = ClassSchool("A")
class_b = ClassSchool("B")

class_a.addStudent("Ana")
class_a.addStudent("Bruno")
class_b.addStudent("Carla")

assert class_a.students == ["Ana", "Bruno"] 
assert class_b.students == ["Carla"] 
assert class_a.students is not class_b.students

assert class_a.numberStudents() == 2 
assert class_b.numberStudents() == 1

ClassSchool.school = "Escola XYZ"

assert class_a.school == "Escola XYZ" 
assert class_b.school == "Escola XYZ"
```
